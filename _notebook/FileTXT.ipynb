## Работа с текстовыми файлами

```
using System.IO;
```

```
const string TEMP_DIR = "temp";

if (!Directory.Exists(TEMP_DIR))
    Directory.CreateDirectory(TEMP_DIR);
```

```
// TEMP_DIR + "\\" + TimeProvider_TXT.txt
Path.Combine(TEMP_DIR, "aaa", "bbb", "TimeProvider_TXT.txt")
```

```
Path.Combine("C:", "aaa\\", "bbb", "TimeProvider_TXT.txt")
```

```
var filename = Path.Combine(TEMP_DIR, "temp_TXT.txt");
Console.WriteLine(filename);

// Запись в файл
var f = new StreamWriter(filename);
try
{
    f.Write("Hello");
    f.WriteLine("123");
    f.WriteLine("Привет");
}
finally
{
    f.Close();
}
```

```
using(var f = new StreamWriter(filename))
{
    f.Write("Hello");
    f.WriteLine("123");
    f.WriteLine("Привет");
}
```
