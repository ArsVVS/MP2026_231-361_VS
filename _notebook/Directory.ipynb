## Directory

```
using System.IO;
```

```
string path = @"C:\Windows\Help";
```

```
Directory.GetFiles(path)
```

```
Directory.GetDirectories(path)
```

```
var r = Directory.GetFiles(path, "", searchOption: SearchOption.AllDirectories);
foreach (var i in r)
    Console.WriteLine(i);
```

```
string.Join(Environment.NewLine, Directory.GetLogicalDrives())
```
