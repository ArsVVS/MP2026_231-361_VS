# Делегат (delegate)

# Пользовательские делегаты

```
delegate void Work();

void Hello() => Console.WriteLine("Hello");
void HowAreYou() => Console.WriteLine("How are you?");

// Однотипный делегат
Work work = Hello;
work = HowAreYou;

// Вызов
work.Invoke();
// Сокращенная форма вызова
work();
```

```
// Многотипный делегат

Work work;
work += Hello;
work += Hello;
work += Hello;
work -= Hello;
work += HowAreYou;
// 
work += delegate () { Console.WriteLine("What?"); };
//
work += () => Console.WriteLine("Вау");

work();
```

```
Work work;
Console.WriteLine(work == null);

if (work != null)
    work.Invoke();

// ?.
work?.Invoke();
```
