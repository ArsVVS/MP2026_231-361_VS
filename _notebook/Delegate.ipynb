# Делегат (delegate)

# Пользовательские делегаты

```
delegate void Work();

void Hello() => Console.WriteLine("Hello");
void HowAreYou() => Console.WriteLine("How are you?");

// Однотипный делегат
Work work = Hello;
work = HowAreYou;

// Вызов
work.Invoke();
// Сокращенная форма вызова
work();
```

```
// Многотипный делегат

Work work;
work += Hello;
work += Hello;
work += Hello;
work -= Hello;
work += HowAreYou;
// 
work += delegate () { Console.WriteLine("What?"); };
//
work += () => Console.WriteLine("Вау");

work();
```

```
Work work;
Console.WriteLine(work == null);

if (work != null)
    work.Invoke();

// ?.
work?.Invoke();
```

```
delegate int MathOperation(int x, int y);

int Add(int x, int y) => x + y;
int Sub(int x, int y) => x - y;
int Mul(int x, int y) => x * y;

MathOperation calc;
calc = Add;

calc(2, 3)
```

```
delegate int MathOperation(int x, int y);

int Add(int x, int y) => x + y;
int Sub(int x, int y) => x - y;
int Mul(int x, int y) => x * y;

Dictionary<string, MathOperation> calc = new()
{
    ["+"] = Add,
    ["-"] = Sub,
    ["*"] = Mul
};

calc["-"](2, 3)
```

```
enum MathOperator {Add, Sub, Mul};
delegate int MathOperation(int x, int y);

int Add(int x, int y) => x + y;
int Sub(int x, int y) => x - y;
int Mul(int x, int y) => x * y;

Dictionary<MathOperator, MathOperation> calc = new()
{
    [MathOperator.Add] = Add,
    [MathOperator.Sub] = Sub,
    [MathOperator.Mul] = Mul
};

calc[MathOperator.Add](2, 3)
```

# Встроенные (стандартные) делегаты

```
// Action (Экшен)

Action printName = () => Console.WriteLine("Привет");
printName();

Action<int, int> printSum = (x, y) => Console.WriteLine(x + y);
printSum += (x, y) => Console.WriteLine($"{x} + {y} = {x + y}");
printSum(2, 3);
```

```
var printName = () => Console.WriteLine("Привет");
printName();

var printSum = (int x, int y) => Console.WriteLine(x + y);
printSum += (x, y) => Console.WriteLine($"{x} + {y} = {x + y}");
printSum(2, 3);
```

```
// Func (Функция делегирования)

Func<string> getName = () => "Ваня";
Console.WriteLine(getName());

Func<int, int, int> getSum = (x, y) => x + y;
Console.WriteLine(getSum(2, 3));
```

```
// Func (Функция делегирования)

var getName = () => "Ваня";
Console.WriteLine(getName());

var getSum = (int x, int y) => x + y;
Console.WriteLine(getSum(2, 3));
```

```
// Predicate (Предикат)
// Принимает только один параметр и возвращает значение типа bool

Predicate<string> isCheckName = (name) => name?.Length > 1;
Console.WriteLine(isCheckName("Ы"));
Console.WriteLine(isCheckName("Ваня"));
Console.WriteLine(isCheckName(null));
```

```
// Примеры

Predicate<int> isPositive = i => i > 0;
Console.WriteLine(isPositive(10));
Console.WriteLine(isPositive(0));
Console.WriteLine(isPositive(-10));

Predicate<int> isEven = i => i % 2 == 0;
Console.WriteLine(isEven(2));
Console.WriteLine(isEven(3));
```

# Событие (event)

```
// Класс с делегатом

class Person
{
    public delegate void MyHandler(string newValue);
    public event MyHandler NameChanged;
    private string name;
    public string Name
    {
        get => name;
        //set => nameof = value;
        //set 
        // {
        //     name = value;
        //     NameChanged?.Invoke(name);
        // }
        set => NameChanged?.Invoke(name = value);
    }
}

var p = new Person();
p.NameChanged += s => Console.WriteLine($"Новое значение = {s}");
p.NameChanged += s => Console.WriteLine($"logs > {s}");
p.Name = "Юра";
p.Name = "Паша";
p.Name = "Максим";
p.NameChanged += s => Console.WriteLine($">>> {s}"); //плохо
p.Name = "Дима";
Console.WriteLine(p.Name);
```
