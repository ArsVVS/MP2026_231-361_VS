# Словарь (Dictionary)

```
Dictionary<int, string> d = new();
d.Add(4, "Хорошо");
d.Add(5, "Отлично");
d.Add(3, "Удовлетворительно");
// Добавить
d[2] = "Неудовлетворительно";
// Редактировать
d[5] = "ОТЛИЧНО";
// Удалить
d.Remove(4);
d
```

```
foreach (var i in d)
    Console.WriteLine(i);
```

```
foreach (var i in d)
    Console.WriteLine($"{i.Key} | {i.Value}");
```

```
string value;
d.TryGetValue(4, out value);
value
```

```
// Проверить наличие ключа или значения
Console.WriteLine(d.ContainsKey(4));
Console.WriteLine(d.ContainsKey(5));
```

```
Dictionary<int, string> d = new()
{
    [4] = "Хорошо",
    [5] = "Отлично",
    [3] = "Удовлетворительно"
};
d
```
