# Список (List)
| Массив | Список |
| - | - |
| фиксированный размер | динамический размер |
| быстрей | удобней |

```
// Не типизированный список (необобщенный список)

ArrayList arrayList = new();
arrayList.Add(3);
arrayList.Add("Привет");
arrayList.Add(3.14);
arrayList
```

```
// Типизированный список (обобщенный список)

List<int> numbers = new();
numbers.Add(1);
numbers.Add(256);
numbers
```
