Функции (Методы)

```
void PrintString ()
{
    Console.WriteLine("Строка");
}
```

```
PrintString();
```

```
using System.ComponentModel.DataAnnotations;

int Sum1(int a, int b)
{
    return a + b;
}

// Позиционные аргументые
Console.WriteLine(Sum1(1, 2));

// Именованные аргументые
Console.WriteLine(Sum1(b:20, a:10));
```

```
int Sum2(int a, int b) => a + b;
Console.WriteLine(Sum2(1, 2));
```

```
// Параметры по умолчанию
// Поминим: Параметры по умолчанию только в конце
// Заполняем справа налево
int Sum3(int a, int b = 10, int c = 100)
{
    return a + b + c;
}
Console.WriteLine(Sum3(1, 2, 3));
Console.WriteLine(Sum3(1));
```

```
void PrintNumbers(int[] numbers)
{
    Console.WriteLine(String.Join(" | ", numbers));
}
PrintNumbers([1, 2, 3, 4, 5]);
```

```
void PrintNumbers2(params int[] numbers)
{
    Console.WriteLine(String.Join(" | ", numbers));
}
PrintNumbers2(1, 2, 3, 4, 5);
```

```
void PrintParams(params object[] numbers)
{
    Console.WriteLine(String.Join(" | ", numbers));
}
PrintParams(1, "Привет", 3.14, 'a');
```

```
// Перегрузка методов
int Add(int a, int b) => a + b;
string Add(string a, string b) => a + b;
Console.WriteLine(Add(1, 2));
Console.WriteLine(Add("1", "2"));
```

```
// Методы расширения
static void Print(this string source)
{
    Console.WriteLine(source);
}
var st = "Строка";
st.Print();
```
