```
using System;


// ВЫВОД
static void PrintShape(string[] shape)
{
    foreach (var line in shape)
        Console.WriteLine(line);
}

// РОМБ (с заливкой)
static string[] GetRhombus(int size, char symbol, bool filled)
{
    string[] result = new string[size];
    int mid = size / 2;

    for (int i = 0; i < size; i++)
    {
        int stars = i <= mid ? 2 * i + 1 : 2 * (size - i - 1) + 1;
        int spaces = (size - stars) / 2;

        string line = new string(' ', spaces);

        for (int j = 0; j < stars; j++)
        {
            if (filled)
            {
                line += symbol;
            }
            else
            {
                // только граница
                if (j == 0 || j == stars - 1)
                    line += symbol;
                else
                    line += ' ';
            }
        }

        result[i] = line;
    }

    return result;
}


// ЗЕБРА (гориз/вертик)
static string[] GetZebra(int width, int height, char symbol, int thickness, bool horizontal)
{
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        string line = "";

        for (int j = 0; j < width; j++)
        {
            bool draw;

            if (horizontal)
                draw = (i / thickness) % 2 == 0;
            else
                draw = (j / thickness) % 2 == 0;

            line += draw ? symbol : ' ';
        }

        result[i] = line;
    }

    return result;
}

// ЗМЕЙКА (нормальная)
static string[] GetSnake(int width, int height, char symbol, bool startLeft)
{
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        char[] line = new char[width];

        // заполняем пробелами
        for (int j = 0; j < width; j++)
            line[j] = ' ';

        if (i % 2 == 0)
        {
            // полная линия
            for (int j = 0; j < width; j++)
                line[j] = symbol;
        }
        else
        {
            // точка справа или слева (чередование)
            if ((i / 2) % 2 == 0)
                line[width - 1] = symbol; // справа
            else
                line[0] = symbol; // слева
        }

        result[i] = new string(line);
    }

    return result;
}

// УЛИТКА (СПИРАЛЬ С ТОЛЩИНОЙ)
static string[] GetSpiral(int width, int height, char symbol, int thickness)
{
    char[,] matrix = new char[height, width];

    // заполняем пробелами
    for (int i = 0; i < height; i++)
        for (int j = 0; j < width; j++)
            matrix[i, j] = ' ';

    int top = 0;
    int bottom = height - 1;
    int left = 0;
    int right = width - 1;

    while (top <= bottom && left <= right)
    {
        // верх 
        for (int j = left; j <= right; j++)
            matrix[top, j] = symbol;
        top++;

        // право 
        for (int i = top; i <= bottom; i++)
            matrix[i, right] = symbol;
        right--;

        // низ 
        if (top <= bottom)
        {
            for (int j = right; j >= left; j--)
                matrix[bottom, j] = symbol;
            bottom--;
        }

        // лево (НО НЕ ДО КОНЦА,  оставляем вход)
        if (left <= right)
        {
            for (int i = bottom; i > top; i--) // ВАЖНО: > top (не >=)
                matrix[i, left] = symbol;
            left++;
        }
    }

    // перевод в строки
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        char[] line = new char[width];

        for (int j = 0; j < width; j++)
            line[j] = matrix[i, j];

        result[i] = new string(line);
    }

    return result;
}
```

```
Console.WriteLine("РОМБ (рамка):");
PrintShape(GetRhombus(7, '*', false));
```
