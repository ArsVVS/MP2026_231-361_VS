```
using System;

// вывод
// эта функция берет массив строк и выводит их по одной строке в консоль
// например: ["***", " * "] выведет:
// ***
//  *
static void PrintShape(string[] shape)
{
    foreach (var line in shape) // перебираем каждую строку
        Console.WriteLine(line); // выводим строку
}

// залитый ромб
// сначала количество звезд увеличивается (верх ромба)
// потом уменьшается (низ ромба)
// перед звездами добавляем пробелы, чтобы выровнять по центру
static string[] GetRhombusFilled(int size, char symbol)
{
    string[] result = new string[size]; // массив строк
    int mid = size / 2; // середина ромба

    for (int i = 0; i < size; i++)
    {
        // считаем сколько символов рисовать в строке
        // до середины увеличиваем, после уменьшаем
        int stars = i <= mid
            ? 2 * i + 1
            : 2 * (size - i - 1) + 1;

        // считаем сколько пробелов нужно слева
        int spaces = (size - stars) / 2;

        // создаем строку: пробелы + символы
        result[i] = new string(' ', spaces) + new string(symbol, stars);
    }

    return result;
}

// ромб контур
// отличие от залитого:
// рисуем только первый и последний символ, середина — пробелы
static string[] GetRhombusBorder(int size, char symbol)
{
    string[] result = new string[size];
    int mid = size / 2;

    for (int i = 0; i < size; i++)
    {
        int stars = i <= mid ? 2 * i + 1 : 2 * (size - i - 1) + 1;
        int spaces = (size - stars) / 2;

        string line = new string(' ', spaces); // сначала пробелы

        for (int j = 0; j < stars; j++)
        {
            // если первый или последний символ — рисуем
            if (j == 0 || j == stars - 1)
                line += symbol;
            else
                line += ' '; // иначе пробел
        }

        result[i] = line;
    }

    return result;
}

// ромб с числами слева
// слева выводим число — сколько символов в строке
static string[] GetRhombusWithNumbers(int size, char symbol)
{
    string[] result = new string[size];
    int mid = size / 2;

    for (int i = 0; i < size; i++)
    {
        int stars = i <= mid ? 2 * i + 1 : 2 * (size - i - 1) + 1;
        int spaces = (size - stars) / 2;

        result[i] =
            $"{stars}".PadLeft(2) + " " + // число слева (с выравниванием)
            new string(' ', spaces) +
            new string(symbol, stars);
    }

    return result;
}

// двойной ромб (пустой внутри)
// сначала заполняем строку символами
// потом "вырезаем" середину пробелами
static string[] GetRhombusDouble(int size, char symbol)
{
    string[] result = new string[size];

    for (int i = 0; i < size; i++)
    {
        int left = i <= size / 2 ? i : size - i - 1;
        int right = size - left - 1;

        char[] line = new char[size];

        // заполняем всю строку символами
        for (int j = 0; j < size; j++)
            line[j] = symbol;

        // заменяем середину на пробелы
        for (int j = left; j <= right; j++)
            line[j] = ' ';

        result[i] = new string(line);
    }

    return result;
}

// зебра с толщиной линий
// делим координату на thickness → получаем полосы
// затем через % 2 делаем чередование
static string[] GetZebra(int width, int height, char symbol, int thickness, bool horizontal)
{
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        string line = "";

        for (int j = 0; j < width; j++)
        {
            bool draw;

            if (horizontal)
                draw = (i / thickness) % 2 == 0; // горизонтальные полосы
            else
                draw = (j / thickness) % 2 == 0; // вертикальные

            line += draw ? symbol : ' ';
        }

        result[i] = line;
    }

    return result;
}

// прямоугольник (залитый или рамка)
// если filled = true полностью заполнен
// иначе рисуем только границу
static string[] GetRectangle(int width, int height, char symbol, bool filled)
{
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        string line = "";

        for (int j = 0; j < width; j++)
        {
            // если залитый всегда рисуем
            // иначе только края
            if (filled || i == 0 || i == height - 1 || j == 0 || j == width - 1)
                line += symbol;
            else
                line += ' ';
        }

        result[i] = line;
    }

    return result;
}

// горизонтальная змейка
// строка-строка-переход
static string[] GetSnakeHorizontal(int width, int height, char symbol, bool reverse)
{
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        char[] line = new char[width];

        // заполняем пробелами
        for (int j = 0; j < width; j++)
            line[j] = ' ';

        if (i % 2 == 0)
        {
            // полностью заполненная строка
            for (int j = 0; j < width; j++)
                line[j] = symbol;
        }
        else
        {
            // точка перехода (слева или справа)
            bool right = ((i / 2) % 2 == 0) ^ reverse;
            line[right ? width - 1 : 0] = symbol;
        }

        result[i] = new string(line);
    }

    return result;
}

// вертикальная змейка
// то же самое, только по столбцам
static string[] GetSnakeVertical(int width, int height, char symbol, bool reverse)
{
    char[,] m = new char[height, width];

    // заполняем пробелами
    for (int i = 0; i < height; i++)
        for (int j = 0; j < width; j++)
            m[i, j] = ' ';

    for (int j = 0; j < width; j++)
    {
        if (j % 2 == 0)
        {
            // полный столбец
            for (int i = 0; i < height; i++)
                m[i, j] = symbol;
        }
        else
        {
            // точка перехода
            bool down = ((j / 2) % 2 == 0) ^ reverse;
            m[down ? height - 1 : 0, j] = symbol;
        }
    }

    // перевод в строки
    string[] result = new string[height];

    for (int i = 0; i < height; i++)
    {
        char[] line = new char[width];

        for (int j = 0; j < width; j++)
            line[j] = m[i, j];

        result[i] = new string(line);
    }

    return result;
}

// залитые песочные часы
// сверху уменьшаем ширину, потом увеличиваем
static string[] GetHourglassFilled(int size, char symbol)
{
    string[] result = new string[size];
    int mid = size / 2;

    for (int i = 0; i < size; i++)
    {
        int spaces = i <= mid ? i : size - i - 1;
        int stars = size - 2 * spaces;

        result[i] = new string(' ', spaces) + new string(symbol, stars);
    }

    return result;
}

// песочные часы линиями 
static string[] GetHourglassLines(int size)
{
    string[] result = new string[size];

    int left = 0;
    int right = size - 1;

    for (int i = 0; i < size; i++)
    {
        char[] line = new char[size];

        // заполняем строку пробелами
        for (int j = 0; j < size; j++)
            line[j] = ' ';

        // рисуем через одну строку
        if (i % 2 == 0)
        {
            if (i < size / 2)
            {
                // верхняя часть
                line[left] = '\\';
                line[right] = '/';
            }
            else
            {
                // нижняя часть (меняем направления)
                line[left] = '/';
                line[right] = '\\';
            }
        }

        result[i] = new string(line);

        // двигаем линии к центру, потом обратно
        if (i < size / 2)
        {
            left++;
            right--;
        }
        else
        {
            left--;
            right++;
        }
    }

    return result;
}
```

```
PrintShape(GetRhombusFilled(9, '*'));
Console.WriteLine("");
PrintShape(GetRhombusBorder(9, '*'));
```

```
PrintShape(GetRhombusWithNumbers(9, '*'));
Console.WriteLine("");
PrintShape(GetRhombusDouble(11, '*'));
```

```
PrintShape(GetZebra(12, 6, '1', 2, true));
Console.WriteLine("");
PrintShape(GetZebra(12, 6, '2', 2, false));
```

```
PrintShape(GetRectangle(12, 5, 'a', true));
Console.WriteLine("");
PrintShape(GetRectangle(12, 5, 'a', false));
```

```
PrintShape(GetSnakeHorizontal(12, 6, '*', false));
Console.WriteLine("");
PrintShape(GetSnakeHorizontal(12, 6, '*', true));
```

```
PrintShape(GetSnakeVertical(8, 6, '.', false));
Console.WriteLine("");
PrintShape(GetSnakeVertical(8, 6, '*', true));
```

```
PrintShape(GetHourglassFilled(7, '4'));
Console.WriteLine("");
PrintShape(GetHourglassLines(7));
```
